# HireMinds AI – Resume & Job Description Matcher

This notebook lets you upload a resume (PDF or TXT) and a job description (TXT) and see the matching result using the trained Random Forest model.

```python
import os, io, sys
import pandas as pd
import ipywidgets as widgets
from IPython.display import display, Markdown

# Ensure the project root is on the path
project_root = os.path.abspath('.'); sys.path.append(project_root)

# Import pipeline functions
from src.pdf_parser import extract_text_from_pdf
from src.preprocessing import clean_text
from src.features import extract_and_build_features
from src.predict import predict_candidate
```

```python
# Helper to load a file (PDF or TXT) and return raw text
def load_resume(file_obj, filename):
    ext = os.path.splitext(filename)[1].lower()
    if ext == '.pdf':
        # Write to a temporary file so pdfplumber can read it
        temp_path = os.path.join('data', 'resumes', filename)
        with open(temp_path, 'wb') as f:
            f.write(file_obj.read())
        return extract_text_from_pdf(temp_path)
    else:
        # Assume plain text
        return file_obj.read().decode('utf-8', errors='ignore')

def load_job(file_obj, filename):
    # Job descriptions are expected to be plain text
    return file_obj.read().decode('utf-8', errors='ignore')
```

```python
# Upload widgets
resume_uploader = widgets.FileUpload(
    accept='.pdf,.txt',
    multiple=False,
    description='Upload Resume'
)
job_uploader = widgets.FileUpload(
    accept='.txt',
    multiple=False,
    description='Upload Job Description'
)
display(resume_uploader)
display(job_uploader)
```

```python
def run_match():
    if not resume_uploader.value or not job_uploader.value:
        print('Please upload both a resume and a job description.')
        return
    # Get uploaded file objects
    resume_key = next(iter(resume_uploader.value))
    job_key = next(iter(job_uploader.value))
    resume_file = resume_uploader.value[resume_key]
    job_file = job_uploader.value[job_key]
    # Load texts
    resume_text = load_resume(io.BytesIO(resume_file['content']), resume_file['metadata']['name'])
    job_text = load_job(io.BytesIO(job_file['content']), job_file['metadata']['name'])
    # Feature extraction
    features, details = extract_and_build_features(resume_text, job_text)
    # Prediction (model file should already exist)
    prediction = predict_candidate(features)
    # Display results nicely
    md = f'''
    ### Prediction Result
    **Candidate Skills:** {', '.join(sorted(details['candidate_skills']))}
    **Required Skills:** {', '.join(sorted(details['required_skills']))}
    **Matched Skills:** {', '.join(sorted(details['matched_skills']))}
    **Missing Skills:** {', '.join(sorted(details['missing_skills']))}
    **Jaccard Similarity:** {features['jaccard_similarity']*100:.2f}%
    **Skill Coverage:** {features['skill_coverage']*100:.2f}%
    **Experience Match:** {features['experience_match']*100:.2f}%
    **Education Match:** {features['education_match']*100:.2f}%
    **Random Forest Prediction:** {prediction['prediction']} (Probability: {prediction['probability']*100:.2f}%)
    '''
    display(Markdown(md))

run_match_button = widgets.Button(description='Run Match', button_style='success')
run_match_button.on_click(lambda b: run_match())
display(run_match_button)
```
